```python
from data_loader import load_blood_cells
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X_raw, y_raw, label_map = load_blood_cells("bloodcells_dataset")

scaler_blood = StandardScaler()
X_scaled_blood = scaler_blood.fit_transform(X_raw)

n_components_blood = 150
pca_blood = PCA(n_components=n_components_blood, random_state=42)
X_blood_pca = pca_blood.fit_transform(X_scaled_blood)

blood_data = {
    "n_categories": len(label_map),
    "n_features": n_components_blood,
    "train_features": X_blood_pca, 
    "train_labels": y_raw,
    "label_names": label_map
}

```
